## <a id='toc1_4_'></a>[详细使用指南](#toc0_)

### <a id='toc1_4_1_'></a>[qibo_profiler_improved.py - Qibo性能分析器](#toc0_)

#### <a id='toc1_4_1_1_'></a>[功能描述](#toc0_)
这是一个高级量子电路性能分析工具，支持多种后端、详细的性能指标收集和自动报告生成。

#### <a id='toc1_4_1_2_'></a>[输入输出说明](#toc0_)

**输入：**
- **必需输入：** Qibo Circuit对象
- **可选输入：**
  - `n_runs`: 运行次数（默认1）
  - `mode`: 分析模式（'basic', 'detailed', 'comprehensive'）
  - `calculate_fidelity`: 是否计算保真度（默认True）
  - `initial_state`: 初始量子态（默认None）
  - `timeout_seconds`: 超时时间（默认300.0秒）

**输出：**
- **返回值：** 包含完整分析结果的字典
  - `metadata`: 分析器版本和时间戳
  - `inputs`: 输入参数和电路属性
  - `results`: 性能指标和统计数据
  - `error`: 错误信息（如果有）
- **生成的文件：** Markdown格式报告（可选）

#### <a id='toc1_4_1_3_'></a>[可变参数详解](#toc0_)

**ProfilerConfig类参数：**
- `n_runs` (int, 默认1): 运行次数，用于统计平均性能
- `mode` (str, 默认'basic'): 分析模式
  - 'basic': 基本性能指标
  - 'detailed': 详细分析，包含更多指标
  - 'comprehensive': 全面分析，包含所有可收集指标
- `calculate_fidelity` (bool, 默认True): 是否计算与基准后端的保真度
- `timeout_seconds` (float, 默认300.0): 单次运行超时时间（秒）
- `version` (str, 默认"1.0"): 分析器版本号

**profile_circuit函数参数：**
- `circuit` (Circuit, 必需): 要分析的Qibo量子电路
- `n_runs` (int, 可选): 运行次数，覆盖配置中的值
- `mode` (str, 可选): 分析模式，覆盖配置中的值
- `calculate_fidelity` (bool, 可选): 是否计算保真度，覆盖配置中的值
- `initial_state` (可选): 初始量子态，默认为|0...0⟩
- `timeout_seconds` (float, 可选): 超时时间，覆盖配置中的值

#### <a id='toc1_4_1_4_'></a>[使用示例](#toc0_)

**基础使用示例：**
```python
from qibo import Circuit, gates
from test.test_function.qibo_profiler_improved import profile_circuit, generate_markdown_report

# 创建测试电路
circuit = Circuit(4)
circuit.add(gates.H(0))
circuit.add(gates.CNOT(0, 1))
circuit.add(gates.RZ(2, theta=0.5))
circuit.add(gates.CNOT(1, 2))
circuit.add(gates.M(0, 1, 2, 3))

# 基本性能分析
result = profile_circuit(circuit, n_runs=10, mode='detailed')
print(f"平均执行时间: {result['results']['summary']['runtime_avg']['value']:.4f}秒")

# 生成详细报告
report_path = generate_markdown_report(result, output_path="my_circuit_report.md")
print(f"报告已保存到: {report_path}")
```

**高级使用示例：**
```python
# 比较不同后端的性能
import qibo
from test.test_function.qibo_profiler_improved import profile_circuit

# 测试电路
circuit = Circuit(6)
for i in range(6):
    circuit.add(gates.H(i))
for i in range(5):
    circuit.add(gates.CNOT(i, i+1))

# 测试不同后端
backends = ["numpy", "qibojit", "qiboml"]
results = {}

for backend in backends:
    try:
        qibo.set_backend(backend)
        result = profile_circuit(circuit, n_runs=5, mode='comprehensive')
        results[backend] = result['results']['summary']['runtime_avg']['value']
        print(f"{backend}: {results[backend]:.4f}秒")
    except Exception as e:
        print(f"{backend} 不可用: {e}")

# 找出最快的后端
fastest_backend = min(results.keys(), key=lambda k: results[k])
print(f"最快后端: {fastest_backend}")
```

#### <a id='toc1_4_1_5_'></a>[最佳实践](#toc0_)
- 对于精确测量，使用 `n_runs >= 5` 获得统计意义
- 大型电路使用 `mode='comprehensive'` 获取完整分析
- 设置合理的 `timeout_seconds` 避免长时间等待
- 使用 `calculate_fidelity=True` 验证计算正确性
- 定期清理缓存以避免内存泄漏

### <a id='toc1_4_3_'></a>[qasmbench_runner_backend_selection.py - QASMBench基准测试](#toc0_)

#### <a id='toc1_4_3_1_'></a>[功能描述](#toc0_)
基于QASMBench标准电路集的基准测试工具，支持多种后端选择和详细的性能比较。

#### <a id='toc1_4_3_2_'></a>[输入输出说明](#toc0_)

**输入：**
- **命令行参数：**
  - `--list`: 列出可用电路
  - `--list-backends`: 列出可用后端
  - `--backend-status`: 显示后端详细状态
  - `--circuit`: 指定QASM文件路径
  - `--backends`: 指定要测试的后端（逗号分隔）
- **必需文件：** QASMBench电路文件（.qasm格式）

**输出：**
- **控制台输出：** 测试进度和结果
- **生成的文件：**
  - CSV报告：`qibobench/reports/{电路名}/benchmark_report.csv`
  - Markdown报告：`qibobench/reports/{电路名}/benchmark_report.md`
  - JSON报告：`qibobench/reports/{电路名}/benchmark_report.json`
  - 电路图：`qibobench/reports/{电路名}/{电路名}_diagram.png`

#### <a id='toc1_4_3_3_'></a>[可变参数详解](#toc0_)

**QASMBenchConfig类参数：**
- `num_runs` (int, 默认5): 每个后端的正式运行次数
- `warmup_runs` (int, 默认1): 预热运行次数（用于JIT编译）
- `output_formats` (list, 默认['csv', 'markdown', 'json']): 输出报告格式
- `baseline_backend` (str, 默认"numpy"): 性能比较基准后端
- `qasm_directory` (str, 默认"../QASMBench"): QASMBench电路根目录

**BackendConfig类参数：**
- `display_name` (str): 后端显示名称（如"qibojit(numba)"）
- `backend_name` (str): Qibo后端名称
- `platform_name` (str, 可选): 平台名称
- `description` (str): 后端描述
- `dependencies` (list): 依赖包列表
- `priority` (int, 默认0): 优先级（用于排序）
- `is_baseline` (bool, 默认False): 是否为基准后端

**命令行参数：**
- `--list`: 列出所有可用电路
- `--list-backends`: 列出所有可用后端
- `--backend-status`: 显示后端详细状态
- `--circuit`: 指定QASM电路文件路径
- `--backends`: 指定要测试的后端（逗号分隔，格式："backend_name"或"backend_name(platform_name)"）

#### <a id='toc1_4_3_4_'></a>[使用示例](#toc0_)

**基础使用示例：**
```bash
# 从项目根目录执行以下命令

# 查看所有可用电路
python qibobench/qasmbench_runner_backend_selection.py --list

# 查看可用后端
python qibobench/qasmbench_runner_backend_selection.py --list-backends

# 测试单个电路
python qibobench/qasmbench_runner_backend_selection.py --circuit QASMBench/small/qft_n4/qft_n4.qasm

# 比较特定后端（注意backend格式：backend_name(platform_name)）
python qibobench/qasmbench_runner_backend_selection.py \
  --circuit QASMBench/medium/qft_n18/qft_n18.qasm \
  --backends "numpy,qibojit(numba),qiboml(jax)"
```

**高级使用示例：**
```bash
# 批量测试多个电路（从项目根目录执行）
circuits=("QASMBench/small/qft_n4/qft_n4.qasm"
          "QASMBench/small/ghz_n4/ghz_n4.qasm"
          "QASMBench/small/grover_n2/grover_n2.qasm")

for circuit in "${circuits[@]}"; do
    echo "测试电路: $circuit"
    python qibobench/qasmbench_runner_backend_selection.py \
        --circuit "$circuit" \
        --backends "numpy,qibojit(numba)"
done
```

#### <a id='toc1_4_3_5_'></a>[最佳实践](#toc0_)
- 优先使用 `transpiled` 版本的QASM文件，避免兼容性问题
- 对于大型电路，考虑只测试关键后端以节省时间
- 使用 `--backend-status` 检查后端依赖是否满足
- 定期清理 `qibobench/reports/` 目录以节省磁盘空间
- 对于性能比较，始终包含基准后端（numpy）

### <a id='toc1_4_4_'></a>[run_benchmarks.py - 量子模拟器基准测试](#toc0_)

#### <a id='toc1_4_4_1_'></a>[功能描述](#toc0_)
多框架量子模拟器基准测试工具，支持多种电路类型、缓存机制和统计分析。

#### <a id='toc1_4_4_2_'></a>[输入输出说明](#toc0_)

**输入：**
- **命令行参数：**
  - `--circuits`: 电路类型列表（默认['qft']）
  - `--qubits`: 量子比特数列表（默认[2,3,4]）
  - `--simulators`: 模拟器列表（默认['qibo-qibojit']）
  - `--golden-standard`: 参考模拟器（默认'qibo-qibojit'）
  - `--output-dir`: 输出目录（默认'results'）
  - `--verbose`: 详细输出模式
  - `--enable-cache`: 启用缓存
  - `--repeat`: 重复运行次数

**输出：**
- **控制台输出：** 测试进度和结果摘要
- **生成的文件：**
  - 时间戳目录：`results/benchmark_{时间戳}/`
  - 性能图表：PNG格式的可视化结果
  - 数据文件：CSV格式的详细数据
  - 摘要报告：Markdown格式的分析报告

#### <a id='toc1_4_4_3_'></a>[可变参数详解](#toc0_)

**命令行参数：**
- `--circuits` (list, 默认["qft"]): 基准测试电路列表
  - 支持: "qft", "grover"
- `--qubits` (list, 默认[2,3,4]): 测试的量子比特数列表
- `--simulators` (list, 默认["qibo-qibojit"]): 模拟器列表
  - 格式: "platform-backend"（例如：qibo-numpy, qibo-qibojit）
- `--golden-standard` (str, 默认"qibo-qibojit"): 参考态生成模拟器
- `--output-dir` (str, 默认"results"): 结果输出目录
- `--verbose`: 启用详细输出模式
- `--repeat` (int, 默认1): 每个电路重复运行次数
- `--warmup-runs` (int, 默认0): 预热运行次数

**缓存相关参数：**
- `--enable-cache` (默认True): 启用参考态缓存
- `--no-cache`: 禁用缓存（覆盖--enable-cache）
- `--cache-type` (默认"hybrid"): 缓存类型
  - "memory": 内存缓存
  - "disk": 磁盘缓存
  - "hybrid": 混合缓存
- `--cache-dir` (默认".benchmark_cache"): 磁盘缓存目录
- `--memory-cache-size` (int, 默认64): 内存缓存最大条目数
- `--clear-cache`: 开始前清空缓存
- `--cache-stats`: 显示缓存统计信息
- `--statistical-analysis`: 启用统计分析

**基础使用示例：**

**高级使用示例：**
```bash
# 从项目根目录执行以下命令

# 完整的性能评估
python Bench/run_benchmarks.py \
  --circuits qft grover \
  --qubits 2 3 4 5 6 7 8 \
  --simulators qibo-numpy qibo-qibojit qiboml-jax \
  --golden-standard qibo-qibojit \
  --repeat 3 \
  --warmup-runs 2 \
  --enable-cache \
  --cache-type hybrid \
  --statistical-analysis \
  --output-dir my_benchmark_results \
  --verbose
```

### <a id='toc1_4_5_'></a>[vqe_bench_new.py - VQE框架性能基准测试](#toc0_)

#### <a id='toc1_4_5_1_'></a>[功能描述](#toc0_)
变分量子本征求解器（VQE）框架性能基准测试工具，支持多框架比较和详细的性能监控。

#### <a id='toc1_4_5_2_'></a>[输入输出说明](#toc0_)

**输入：**
- **配置文件：** 需要vqe_config.py配置文件
- **可选参数：** 通过配置文件设置
  - 框架列表：['Qiskit', 'PennyLane', 'Qibo']
  - 量子比特数范围：[4, 6, 8]
  - 优化器配置：COBYLA, SPSA, L-BFGS-B
  - 问题参数：J耦合、h场等

**输出：**
- **控制台输出：** 测试进度和统计摘要
- **返回值：** 包含所有结果的字典
- **生成的文件：**
  - JSON结果文件：`{实验名}_{时间戳}.json`

#### <a id='toc1_4_5_3_'></a>[可变参数详解](#toc0_)

**配置参数（通过vqe_config.py设置）：**

**框架配置：**
- `frameworks_to_test` (list): 要测试的框架列表

- `backend_details` (dict): 各框架的后端配置

**问题配置：**

- `ansatz_type` (str, 默认"HardwareEfficient"): Ansatz类型

**优化器配置：**
- `optimizer` (str, 默认"COBYLA"): 优化器类型
  - "COBYLA": 约束优化
  - "SPSA": 同时扰动随机近似
  - "L-BFGS-B": 有限内存BFGS
- `max_evaluations` (int, 默认1000): 最大函数评估次数
- `accuracy_threshold` (float, 默认1e-4): 收敛阈值

**系统配置：**
- `n_qubits_range` (list): 测试的量子比特数范围
- `n_runs` (int, 默认3): 每个配置的重复运行次数
- `max_time_seconds` (int, 默认1800): 单次运行最大时间
- `max_memory_mb` (int, 默认4096): 最大内存使用限制
- `output_dir` (str): 结果输出目录
- `save_results` (bool): 是否保存结果到文件

### <a id='toc1_5_2_'></a>[Q1: 如何安装所需的依赖包？](#toc0_)

### <a id='toc1_5_3_'></a>[Q2: 脚本运行时出现内存不足错误怎么办？](#toc0_)

**A:** 可以尝试以下解决方案：
1. 减少测试的量子比特数
2. 降低重复运行次数
3. 启用缓存机制减少重复计算
4. 增加系统虚拟内存
5. 使用更高效的后端（如qibojit）

### <a id='toc1_5_4_'></a>[Q3: 如何选择合适的后端？](#toc0_)

**A:** 根据需求选择：
- **numpy**: 最稳定，适合小型电路和调试
- **qibojit**: 性能最佳，适合中型电路
- **qiboml**: 适合与机器学习结合的应用
- **qibotn**: 适合大型稀疏电路

### <a id='toc1_5_5_'></a>[Q4: 如何解释性能分析结果？](#toc0_)

**A:** 关键指标说明：
- **runtime_avg**: 平均执行时间，越小越好
- **memory_usage_avg**: 平均内存使用，越小越好
- **fidelity**: 计算保真度，越接近1越好
- **cpu_utilization**: CPU利用率，反映资源使用效率

### <a id='toc1_5_6_'></a>[Q5: 为什么脚本运行时出现ImportError或ModuleNotFoundError？](#toc0_)

**A:** 常见原因和解决方案：

2. **Python路径问题：** 对于VQE测试，可以尝试：
   ```bash
   cd Bench/VQEtest && python vqe_bench_new.py && cd ../..
   # 或者
   python -c "import sys; sys.path.append('Bench/VQEtest'); import vqe_bench_new; vqe_bench_new.main()"
   ```

3. **模块未安装：** 检查是否已安装所有必要的依赖包

### <a id='toc1_5_7_'></a>[Q6: 如何理解和使用不同脚本中的参数格式？](#toc0_)

**A:** 各脚本的参数格式说明：

   - 默认值为["qibo-qibojit"]
